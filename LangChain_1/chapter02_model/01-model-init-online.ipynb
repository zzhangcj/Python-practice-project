### 1.1 调用DeepSeek官网的DeepSeek模型

调用大模型涉及到的三个特别重要的参数：
base_url、model_name、api_key

举例1:将参数信息存储在.env配置文件中

```python
from langchain.chat_models import init_chat_model
from langchain_community.chat_models import ChatZhipuAI, ChatTongyi
from langchain_deepseek import ChatDeepSeek
from dotenv import load_dotenv
import os

#读取.env配置文件中的信息，相关的环境变量以.env文件中的优先
load_dotenv(verbose=True)

DEEPSEEK_API_KEY=os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL=os.getenv("DEEPSEEK_BASE_URL")

#2.模型的初始化
llm_deepseek=ChatDeepSeek(
    model="deepseek-v4-flash",
    api_key=DEEPSEEK_API_KEY,
    api_base=DEEPSEEK_BASE_URL,

)

#3.模型的调用
response=llm_deepseek.invoke("请用一句话介绍自己")
print(response)
```

## 4.使用LangChain1.x的统一方式

使用 init_chat_model()

举例1：调用DeepSeek官网的DeepSeek模型

```python
from dotenv import load_dotenv
import os

#读取.env配置文件中的信息，相关的环境变量以.env文件中的优先
load_dotenv(verbose=True)

DEEPSEEK_API_KEY=os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL=os.getenv("DEEPSEEK_BASE_URL")

#2.获取大模型
model=init_chat_model(
    # model="deepseek-v4-flash",
    # model_provider="deepseek",
    #等价于
    model="deepseek:deepseek-v4-flash",
    api_key=DEEPSEEK_API_KEY,
    base_url=DEEPSEEK_BASE_URL,
)

#3.模型的调用
print(model.invoke("请用一句话介绍自己"))
```

举例2：调用CloseAI

```python
from dotenv import load_dotenv
import os

#读取.env配置文件中的信息，相关的环境变量以.env文件中的优先
load_dotenv(verbose=True)

CLOSEAI_API_KEY=os.getenv("DEEPSEEK_API_KEY")
CLOSEAI_BASE_URL=os.getenv("DEEPSEEK_BASE_URL")

#2.获取大模型
model=init_chat_model(
    model="openai:deepseek-v4-flash",
    api_key=CLOSEAI_API_KEY,
    base_url=CLOSEAI_BASE_URL,
)

#3.模型的调用
print(model.invoke("请用一句话介绍自己"))
```

举例3：模型换成gpt

```python
from dotenv import load_dotenv
import os

#读取.env配置文件中的信息，相关的环境变量以.env文件中的优先
load_dotenv(verbose=True)

CLOSEAI_API_KEY=os.getenv("DEEPSEEK_API_KEY")
CLOSEAI_BASE_URL=os.getenv("DEEPSEEK_BASE_URL")

#2.获取大模型
model=init_chat_model(
    model="openai:gpt-5-mini",
    api_key=CLOSEAI_API_KEY,
    base_url=CLOSEAI_BASE_URL,
)

#3.模型的调用
print(model.invoke("请用一句话介绍自己"))
```
