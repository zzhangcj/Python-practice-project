# 使用本地ollama中的模型进行初始化

举例1：使用ChatOllama

```python
from langchain_ollama import ChatOllama
ollama_llm = ChatOllama(
model="deepseek-r1:1.5b",
base_url="http://192.168.1.106:11434"
#如果Ollama在本地默认端口运行，则可省略，或使用http://localhost:11434
)
question = "你好，请你介绍一下你自己。"
result = ollama_llm.invoke(question)
print(result)
```

举例2：使用init_chat_model()

```python
from langchain.chat_models import init_chat_model
ollama_llm = init_chat_model(
model="deepseek-r1:1.5b",
model_provider="ollama", #这里比较重要，不写的话会报错，因为默认为deepseek
# base_url="http://192.168.1.106:11434",
)
question = "你好，请你介绍一下你自己。"
result = ollama_llm.invoke(question)
print(result)
```
